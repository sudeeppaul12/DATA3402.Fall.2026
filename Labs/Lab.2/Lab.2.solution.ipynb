*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
import copy 

def computer_player(board, computer_player):
    n = len(board)

    # Since both can't be a computer, the alternate player is human
    if computer_player == 1:
        human = 2
    else:
        human = 1

    # Go for the win
    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:
                new_board = copy.deepcopy(board)
                make_move(new_board, computer_player, i, j)

                if check_outcome(new_board) == computer_player:
                    make_move(board, computer_player, i, j)
                    return True

    # Go for the block
    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:
                new_board = copy.deepcopy(board)
                make_move(new_board, human, i, j)

                if check_outcome(new_board) == human:
                    make_move(board, computer_player, i, j)
                    return True

    # Otherwise, pick the first available spot
    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:
                make_move(board, computer_player, i, j)
                return True

    return False
````

````python
def tree_search(scores, maximizing):
    

    if not isinstance(scores, list):
        return scores

    if len(scores) == 0:
        return 0

    results = []

    for score in scores:
        results.append(
            tree_search(score, not maximizing)
        )

    if maximizing:
        return max(results)

    else:
        return min(results)
````

````python
def pick_move(board, computer, depth):

    moves, scores = generate_moves(
        board,
        computer,
        computer,
        depth
    )

    if len(moves) == 0:
        return None

    results = []

    for score in scores:
        results.append(
            tree_search(score, False)
        )

    best_score = max(results)
    best_index = results.index(best_score)

    return moves[best_index]
````
